## Perfil profesional

Data Analyst / BI Analyst con experiencia en análisis de datos operativos y de negocio, desarrollo de dashboards, reporting y soluciones de Business Intelligence. En Vueling Airlines he trabajado en proyectos de análisis de datos aplicados al mantenimiento aeronáutico, desarrollando herramientas de seguimiento de KPIs, análisis del rendimiento de fuselaje y motores e identificación de incidencias y patrones recurrentes. Experiencia colaborando con equipos técnicos y de operaciones para transformar datos en información útil para la toma de decisiones y la optimización de procesos.

Cuento con conocimientos en SQL, modelado y explotación de datos, así como experiencia en visualización con Tableau, Power BI y Looker Studio, y herramientas como Snowflake y MicroStrategy. Complemento mi perfil con conocimientos de Python aplicado al análisis de datos y Machine Learning (Pandas, NumPy y Scikit-learn), además de experiencia previa en analítica digital y marketing, trabajando con KPIs, tráfico web y resultados de campañas. Destaco por mi pensamiento analítico, orientación al detalle y capacidad para conectar las necesidades de negocio con soluciones basadas en datos.

## Charlas y eventos 


### Ponente – «Del Marketing Digital a la explotación del Big Data»

**IES Virgen de las Nieves | Granada**  
*Ciclo Formativo de Hostelería y Turismo*

Diseño y presentación de una charla formativa dirigida al alumnado de Formación Profesional de Hostelería y Turismo, abordando la evolución del marketing digital hacia la gestión, explotación y análisis de datos en el entorno empresarial.

**Contenidos impartidos:**

- **Marketing Digital:** estrategias y acciones de marketing digital, publicidad en redes sociales, email marketing y generación de contenido orgánico.
- **Bases de datos:** importancia de los datos en la empresa, organización, gestión y explotación de bases de datos.
- **Big Data:** introducción al concepto, características y aplicaciones en el entorno empresarial.
- **Business Intelligence y Business Analytics:** herramientas y metodologías para transformar datos en información útil para la toma de decisiones.
- **Medición y análisis:** interpretación de métricas de redes sociales, analítica digital y visualización de indicadores mediante dashboards.
- **Orientación profesional:** presentación de los principales perfiles laborales vinculados al marketing digital, Business Intelligence y ecosistema de datos.

**Competencias destacadas:** comunicación, divulgación de conocimientos técnicos, capacidad didáctica y conexión entre tecnología, análisis de datos y aplicaciones empresariales.
