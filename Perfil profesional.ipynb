## Perfil profesional

Data Analyst / BI Analyst con experiencia en análisis de datos operativos y de negocio, desarrollo de soluciones de Business Intelligence, visualización y reporting. Especializada en transformar datos complejos en indicadores, análisis y herramientas de seguimiento que faciliten la toma de decisiones y contribuyan a mejorar los procesos empresariales.

He desarrollado mi experiencia en el sector aeronáutico, trabajando en Vueling Airlines en proyectos de análisis de datos aplicados al mantenimiento de flota. Mi trabajo ha estado orientado al desarrollo de dashboards para el seguimiento de KPIs operativos, el análisis del rendimiento de fuselaje y motores, la identificación de incidencias recurrentes y la detección de patrones de comportamiento. Esta experiencia me ha permitido adquirir una comprensión práctica de los procesos de mantenimiento aeronáutico y de la importancia de la calidad y fiabilidad de los datos en entornos operativos exigentes.

Cuento con experiencia en modelado y explotación de datos mediante SQL, así como en el desarrollo de soluciones de visualización y reporting con Tableau, Power BI y Looker Studio. También he trabajado con herramientas y tecnologías como Snowflake y MicroStrategy, y he ampliado mis conocimientos en Python aplicado al análisis de datos y Data Science, utilizando bibliotecas como Pandas, NumPy y Scikit-learn.

Mi trayectoria profesional incluye también experiencia en comunicación corporativa y marketing digital, donde he trabajado en análisis de KPIs, estadísticas, tráfico web, resultados de campañas y rendimiento de acciones digitales. Esta etapa me ha proporcionado una base sólida en interpretación de métricas, análisis del comportamiento de usuarios y comunicación de resultados.

Me caracterizo por mi pensamiento analítico, capacidad de organización, orientación al detalle y facilidad para comprender procesos y trasladar necesidades de negocio a soluciones basadas en datos. Combino una visión técnica con una perspectiva funcional, prestando especial atención a que los análisis y herramientas desarrollados sean útiles, comprensibles y estén alineados con los objetivos de cada área.
